```python

import numpy as np

# Creating matrices using the newer Generator API
rng = np.random.default_rng()
A = rng.integers(1, 10, size=(3, 3))
B = rng.integers(1, 10, size=(3, 3))

# 1. Addition using the explicit function
# This is equivalent to A + B
matrix_sum = np.add(A, B)

# 2. Multiplication (Dot Product) using np.dot
# This is equivalent to A @ B
matrix_product = np.dot(A, B)

# 3. BONUS: Element-wise Multiplication
# This multiplies corresponding positions only
element_prod = np.multiply(A, B)
```
