```python

import numpy as np

# Creating matrices using the newer Generator API
rng = np.random.default_rng()
A = rng.integers(1, 10, size=(3, 3))
B = rng.integers(1, 10, size=(3, 3))

# 1. Addition using the explicit function
# This is equivalent to A + B
matrix_sum = np.add(A, B)

# 2. Multiplication (Dot Product) using np.dot
# This is equivalent to A @ B
matrix_product = np.dot(A, B)

# 3. BONUS: Element-wise Multiplication
# This multiplies corresponding positions only
element_prod = np.multiply(A, B)
```

```python
import pandas as pd
import numpy as np

# 1. Create a sample DataFrame
data = {
    'Full_Name': ['ALICE smith', 'bob JOHNSON', 'charlie BROWN', 'DANA white'],
    'Age': [25, 17, 30, 15],
    'City': ['New York', 'London', 'new york', 'Paris'],
    'Score': [88, 42, 95, 60]
}

df = pd.DataFrame(data)
print("--- Original DataFrame ---")
print(df)

# 2. Rename columns (Lowercase the headers)
df = df.rename(columns={'Full_Name': 'name', 'Age': 'age', 'City': 'city', 'Score': 'score'})

# 3. Case Conversion (Normalize the 'name' column)
df['name'] = df['name'].str.title()

# 4. Replace values (Fixing inconsistent city names)
df['city'] = df['city'].replace('new york', 'New York')

# 5. Condition Function (Assign 'Status' based on age)
df['status'] = np.where(df['age'] >= 18, 'Adult', 'Minor')

# 6. Sorting (Shorts - Sort by score descending)
df = df.sort_values(by='score', ascending=False)

print("\n--- Processed DataFrame ---")
print(df)
```
