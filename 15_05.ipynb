```python
'''using insert in a sample dataframe'''
# Sample DataFrame
df = pd.DataFrame({
    'Name': ['Yadu', 'Rithu', 'Muth'],
    'Score': [99, 95, 97]
})

# Insert 'City' at index 1 (between Name and Score)
df.insert(1, 'City', ['Chennai', 'Bangalore', 'Chennai'])
print(df)
```

```python
'''using delete in a sample dataframe'''
# Delete the 'City' column
df_dropped = df.drop('City', axis=1)

# Delete the first row (index 0)
df_no_row = df.drop(0, axis=0)

print(df_dropped)
```

```python
'''using assign'''
# Create a 'Passed' column based on Score
df_new = df.assign(Passed = df['Score'] > 90)
print(df_new)
```

```python
'''using replace'''
# Replace 'Chennai' with 'Madras'
df_replaced = df.replace('Chennai', 'Madras')

# Replace multiple values using a dictionary
df_cleaned = df.replace({99: 100, 'Bangalore': 'Bengaluru'})
print(df_cleaned)
```
