```python
import pandas as pd, numpy as np,matplotlib.pyplot as plt, seaborn as sns
```

```python
data = {
    'Roll No': [1, 2, 3, 4, 5, 6],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank'],
    'Marks': [75, 45, 88, 62, 30, 95]
}

df_students = pd.DataFrame(data)

# Define a passing threshold
passing_threshold = 50

# Determine pass or fail
df_students['Pass/Fail'] = np.where(df_students['Marks'] >= passing_threshold, 'Pass', 'Fail')

# Display the DataFrame
print("Student Data:")
display(df_students)
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='Pass/Fail', y='Marks', hue='Pass/Fail', data=df_students, palette={'Pass': 'green', 'Fail': 'red'})
plt.title('Student Marks Distribution by Pass/Fail Status')
plt.xlabel('Student Name')
plt.ylabel('Marks')
plt.legend(title='Status')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.countplot(x='Pass/Fail', data=df_students, palette={'Pass': 'green', 'Fail': 'red'}, hue='Pass/Fail', legend=False)
plt.title('Number of Students Passed vs. Failed')
plt.xlabel('Pass/Fail Status')
plt.ylabel('Number of Students')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='Name', y='Marks', hue='Pass/Fail', data=df_students, palette={'Pass': 'green', 'Fail': 'red'}, dodge=False)
plt.title('Individual Student Marks Comparison by Pass/Fail Status')
plt.xlabel('Student Name')
plt.ylabel('Marks')
plt.ylim(0, 100) # Assuming marks are out of 100
plt.axhline(y=passing_threshold, color='blue', linestyle='--', label='Passing Threshold') # Add passing threshold line
plt.legend(title='Status', loc='upper left')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(7, 5))

# Draw the distribution boxes
sns.boxplot(x='Pass/Fail', y='Marks', data=df_students, palette={'Pass': '#2ecc71', 'Fail': '#e74c3c'}, width=0.4, hue='Pass/Fail', legend=False)

# Overlay the actual student data points so you can see where they sit
sns.stripplot(x='Pass/Fail', y='Marks', data=df_students, color='black', size=8, jitter=0.1)

plt.title('Distribution Range: Passed vs. Failed Marks', fontsize=14, pad=15)
plt.xlabel('Status', fontsize=12)
plt.ylabel('Marks Range', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
plt.figure(figsize=(7, 5))

# Create the swarm plot
# size=10 makes the individual student points larger and easier to see
sns.swarmplot(
    x='Pass/Fail',
    y='Marks',
    data=df_students,
    hue='Pass/Fail',
    palette={'Pass': '#2ecc71', 'Fail': '#e74c3c'},
    size=10,
    legend=False
)

# Add the target threshold line for clear reference
plt.axhline(y=50, color='blue', linestyle='--', linewidth=1.5, label='Passing Threshold (50)')

# Formatting the graph
plt.title('Individual Student Marks Distribution (Swarm Plot)', fontsize=14, pad=15)
plt.xlabel('Status', fontsize=12)
plt.ylabel('Marks', fontsize=12)
plt.ylim(0, 100)  # Marks range up to 100
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.legend(loc='upper left')

plt.show()
```

```python
plt.figure(figsize=(7, 5))

# Create the strip plot
# jitter=0.1 adds just enough horizontal randomness to see closely packed points
# size=9 makes the student markers prominent
sns.stripplot(
    x='Pass/Fail',
    y='Marks',
    data=df_students,
    hue='Pass/Fail',
    palette={'Pass': '#2ecc71', 'Fail': '#e74c3c'},
    size=9,
    jitter=0.1,
    legend=False
)

# Add the target threshold line for clear reference
plt.axhline(y=50, color='blue', linestyle='--', linewidth=1.5, label='Passing Threshold (50)')

# Formatting the graph
plt.title('Individual Student Marks Distribution (Strip Plot)', fontsize=14, pad=15)
plt.xlabel('Status', fontsize=12)
plt.ylabel('Marks', fontsize=12)
plt.ylim(0, 100)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.legend(loc='upper left')

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Create the figure-level categorical plot
# kind='swarm' handles the layout, and we can still use hue for coloring
g = sns.catplot(
    x='Pass/Fail',
    y='Marks',
    hue='Pass/Fail',
    data=df_students,
    kind='swarm',
    palette={'Pass': '#2ecc71', 'Fail': '#e74c3c'},
    s=10,             # Marker size (equivalent to 'size' in swarmplot)
    height=5,         # Height of the figure in inches
    aspect=1.2        # Aspect ratio (width = height * aspect)
)

# Since catplot returns a FacetGrid, we add a threshold line to its underlying axes
for ax in g.axes.flat:
    ax.axhline(y=50, color='blue', linestyle='--', linewidth=1.5, label='Passing Threshold (50)')
    ax.set_ylim(0, 100)
    ax.grid(axis='y', linestyle='--', alpha=0.5)

# Final formatting adjustments
g.set_axis_labels("Status", "Marks", fontsize=12)
plt.title('Student Marks Comparison using catplot()', fontsize=14, pad=20)
plt.show()
```

```python
plt.figure(figsize=(7, 5))

# Create the violin plot
# inner='point' draws the actual student data points inside the violin
sns.violinplot(
    x='Pass/Fail',
    y='Marks',
    data=df_students,
    hue='Pass/Fail',
    palette={'Pass': '#2ecc71', 'Fail': '#e74c3c'},
    inner='point',  # Options: 'point', 'stick', 'box', None
    density_norm='count', # Adjusts violin width based on the number of students in that group
    legend=False
)

# Add your passing threshold line for quick comparison
plt.axhline(y=50, color='blue', linestyle='--', linewidth=1.5, label='Passing Threshold (50)')

# Formatting the graph
plt.title('Distribution Shape: Passed vs. Failed Marks (Violin Plot)', fontsize=14, pad=15)
plt.xlabel('Status', fontsize=12)
plt.ylabel('Marks', fontsize=12)
plt.ylim(0, 100)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.legend(loc='upper left')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

# Create the scatter plot
# s=120 makes the markers larger and easier to read
sns.scatterplot(
    x='Roll No',
    y='Marks',
    hue='Pass/Fail',
    data=df_students,
    palette={'Pass': '#2ecc71', 'Fail': '#e74c3c'},
    s=120,
    style='Pass/Fail', # Gives different shapes (circle/cross) to Pass vs Fail
)

# Add the target threshold line for clear reference
plt.axhline(y=50, color='blue', linestyle='--', linewidth=1.5, label='Passing Threshold (50)')

# Text annotations: Label each point with the student's name
for i in range(len(df_students)):
    plt.text(
        x=df_students['Roll No'][i] + 0.1,  # Offset slightly to the right of the dot
        y=df_students['Marks'][i] - 1,      # Offset slightly down
        s=df_students['Name'][i],
        fontsize=10,
        verticalalignment='center'
    )

# Formatting the graph
plt.title('Student Marks vs. Roll Number Comparison', fontsize=14, pad=15)
plt.xlabel('Roll Number', fontsize=12)
plt.ylabel('Marks', fontsize=12)
plt.xlim(0, 7)  # Give breathing room for the text labels
plt.ylim(0, 100)
plt.xticks(df_students['Roll No']) # Ensure only actual Roll Numbers show as ticks
plt.grid(axis='both', linestyle='--', alpha=0.4)
plt.legend(loc='upper left')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

# Create the comparison KDE plot
# common_norm=False ensures both curves are normalized independently
# fill=True shades the area under the curves for clear visual overlap comparison
sns.kdeplot(
    x='Marks',
    hue='Pass/Fail',
    data=df_students,
    palette={'Pass': '#2ecc71', 'Fail': '#e74c3c'},
    fill=True,
    alpha=0.4,
    linewidth=2,
    common_norm=False
)

# Add your passing threshold line for reference
plt.axvline(x=50, color='blue', linestyle='--', linewidth=1.5, label='Passing Threshold (50)')

# Formatting the graph
plt.title('Density Comparison: Passed vs. Failed Marks (KDE Plot)', fontsize=14, pad=15)
plt.xlabel('Marks', fontsize=12)
plt.ylabel('Density', fontsize=12)
plt.xlim(0, 100) # Boundary lines for a realistic marks range
plt.grid(axis='both', linestyle='--', alpha=0.3)
plt.legend(title='Status', loc='upper left')

plt.show()
```

```python
plt.figure(figsize=(7, 4.5))

# Comparing the volume of teens per platform
sns.countplot(
    x='platform_usage',
    data=df,
    palette='Set2',
    order=df['platform_usage'].value_counts().index # Orders bars from highest to lowest count
)

plt.title('Distribution of Primary Social Media Platforms', fontsize=14, pad=15)
plt.xlabel('Platform', fontsize=12)
plt.ylabel('Number of Teens', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.show()
```

```python
# Bar plot comparing average stress levels
sns.catplot(x='platform_usage', y='stress_level', hue='gender', kind='bar', data=df, aspect=1.5)
```

```python
# Split violin comparing academic performance across platforms by gender
sns.violinplot(
    x='platform_usage',
    y='academic_performance',
    hue='gender',
    split=True,  # Merges the two hue groups into a single split violin
    data=df,
    palette='muted'
)
plt.title('Academic Performance Distribution by Platform & Gender')
plt.show()
```
